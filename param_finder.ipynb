```python
import pandas as pd

df = pd.read_csv("mining_results_adj.csv")

data_cols = [col for col in df.columns if col.startswith('LOSS_')]

results = []

for col in data_cols:
    # Get the 3 rows with the smallest values for the current column
    top_3_rows = df.nsmallest(3, col)
    
    # Iterate through those 3 rows to extract data
    for index, row in top_3_rows.iterrows():
        results.append({
            'Column': col,
            'Min Value': row[col],
            'id_1': row['source_db'],
            'id_2': row['trial_id']
        })

# 4. View the results
results_df = pd.DataFrame(results)
print(results_df)
```

```python
results_df.to_csv("lowest_losses_full_adj.csv")
```
